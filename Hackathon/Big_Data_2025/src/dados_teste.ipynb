### Tranfomando as BDs PARQUET em XLSX, para entender melhor os dados

```python
import pandas as pd

# Caminho do arquivo Parquet
arquivo_transacoes = '../data/transacoes_2022.parquet'
arquivo_produtos = '../data/produtos.parquet'
arquivo_pdv = '../data/pdvs.parquet'

# Ler arquivo parquet
df1 = pd.read_parquet(arquivo_transacoes)
df2 = pd.read_parquet(arquivo_produtos)
df3 = pd.read_parquet(arquivo_pdv)

# Selecionar apnas as 30 primeiras linhas para amostra
df1_amostra = df1.head(31)


# Exportar para Excel (XLSX)
arquivo_excel_transacoes = '../data/dados_transacoes.xlsx'
df1_amostra.to_excel(arquivo_excel_transacoes, index=False)  # index=False para não incluir a coluna do índice

arquivo_excel_produtos = '../data/dados_produtos.xlsx'
df2.to_excel(arquivo_excel_produtos, index=False)  # index=False para não incluir a coluna do índice

arquivo_excel_pdv = '../data/dados_pdvs.xlsx'
df3.to_excel(arquivo_excel_pdv, index=False)  # index=False para não incluir a coluna do índice

print(f"Arquivos convertidos com sucesso: {arquivo_excel_transacoes}, {arquivo_excel_produtos} e {arquivo_excel_pdv}")
```
