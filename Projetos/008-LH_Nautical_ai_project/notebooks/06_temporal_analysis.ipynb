```python
import pandas as pd

vendas = pd.read_csv('../data/raw/vendas_2023_2024.csv')

vendas['sale_date'] = pd.to_datetime(
    vendas['sale_date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)

# =========================
# Calendário completo
# =========================
datas = pd.date_range(
    start=vendas['sale_date'].min(),
    end=vendas['sale_date'].max(),
    freq='D'
)

calendario = pd.DataFrame({'data': datas})

# =========================
# Vendas diárias
# =========================
vendas_diarias = (
    vendas.groupby(vendas['sale_date'].dt.date)['total']
    .sum()
    .reset_index()
)

vendas_diarias.columns = ['data', 'valor_venda']
vendas_diarias['data'] = pd.to_datetime(vendas_diarias['data'])

# =========================
# Merge
# =========================
df = calendario.merge(vendas_diarias, on='data', how='left')

df['valor_venda'] = df['valor_venda'].fillna(0)

# =========================
# Dia da semana
# =========================
dias = {
    0: 'Segunda-feira',
    1: 'Terça-feira',
    2: 'Quarta-feira',
    3: 'Quinta-feira',
    4: 'Sexta-feira',
    5: 'Sábado',
    6: 'Domingo'
}

df['dia_semana'] = df['data'].dt.weekday.map(dias)

# =========================
# Resultado ordenado
# =========================
resultado = (
    df.groupby('dia_semana')['valor_venda']
    .mean()
    .sort_values()
)

# =========================
# FORMATAÇÃO
# =========================
resultado_formatado = resultado.apply(lambda x: f"R$ {x:,.2f}")

print(resultado_formatado)
```

```python
resultado_df = resultado.reset_index()
resultado_df.columns = ['dia_semana', 'media_vendas']

resultado_df['media_vendas_formatado'] = resultado_df['media_vendas'].apply(
    lambda x: f"R$ {x:,.2f}"
)

resultado_df
```
